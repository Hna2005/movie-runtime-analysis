# data cleaning

## Import Libraries

```python
import ast
import pandas as pd
import numpy as np
```

## Load Raw Data

```python
df = pd.read_csv("../data/raw.csv", low_memory=False)
df.head()
```

```python
df.info()
df.describe(include="all")
```

## Missing Values Analysis
Drop columns with high missing values or low relevance
Columns with more than 50% missing or those considered irrelevant for our analysis are removed.

```python
cols_zero = ["budget", "revenue", "runtime"]

# to numeric & zero to NaN
for col in cols_zero:
    df[col] = pd.to_numeric(df[col], errors='coerce')
    df[col] = df[col].replace(0, np.nan)

# Percentage of missing data and sort
missing_precent = df.isnull().mean() * 100
missing_precent = missing_precent.sort_values(ascending=False)

# Columns with more then 50% missing data
high_missing_cols = missing_precent[missing_precent > 50].index.to_list()

# useless cols
useless_cols = ["homepage", "poster_path", "tagline", "belongs_to_collection", "adult", "video", "id", "overview", "imdb_id"]

# merge cols
drop_cols = list(set(high_missing_cols + useless_cols))

# drop
df = df.drop(columns=drop_cols)
df = df.dropna(subset=['runtime'])

df.info()
df.isnull().sum()
```

## Data type transformation

```python
# Time type
df["release_date"] = pd.to_datetime(df["release_date"], errors="coerce")
df["release_date"] = df["release_date"].dt.year

# Transform JSON to list
json_cols = ["genres", "production_companies", "production_countries", "spoken_languages"]
for j_col in json_cols:
    df[j_col] = df[j_col].apply(
        lambda x: [d['name'] for d in ast.literal_eval(x)] 
        if isinstance(x, str) else []
)
```

## save the clean data

```python
df.to_csv("../data/processed/clean_code.csv", index=False)
```
