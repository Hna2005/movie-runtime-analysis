## Data Quality Checks

```python
# Boolean
print(data['video'].value_counts())
print(f"\n{data['adult'].value_counts()}")
```

```python
# Categorical - Validate JSON columns
def analyze_non_dict_columns(df, columns_to_check):
    results = []
    for col in columns_to_check:
        non_empty = df[col][df[col].notna() & (df[col].astype(str).str.strip() != '')]
        invalid = non_empty[~non_empty.apply(lambda x: isinstance(x, list))]
        
        total = len(non_empty)
        count = len(invalid)
        results.append({
            'column': col,
            'total_non_empty': total,
            'invalid_count': count,
            'invalid_percent': round((count / total * 100), 2) if total > 0 else 0
        })
    return pd.DataFrame(results)

columns_to_check = ['belongs_to_collection', 'genres', 'production_companies', 
                    'production_countries', 'spoken_languages']
result_df = analyze_non_dict_columns(data, columns_to_check)
print(result_df)
```

```python
# Categorical
def analyze_non_dict_columns(df, columns_to_check):
    results = []
    for col in columns_to_check:
        non_empty = df[col][df[col].notna() & (df[col].astype(str).str.strip() != '')]
        not_dict = non_empty[~non_empty.apply(lambda x: isinstance(x, str))]
        total = len(non_empty)
        count = len(not_dict)
        results.append({
            'column': col,
            'total_non_empty': total,
            'non_dict_count': count,
            'non_dict_percent': round((count / total * 100), 2) if total > 0 else 0
        })
    return pd.DataFrame(results)

columns_to_check = ['belongs_to_collection', 'genres', 'production_companies', 
                    'production_countries', 'spoken_languages']
result_df = analyze_non_dict_columns(data, columns_to_check)
print(result_df)
```

```python
# Date
total = data["release_date"].notna().sum()
valid_dates = pd.to_datetime(data["release_date"], errors='coerce').notna().sum()
invalid_count = total - valid_dates
print(f"invalid dates: {invalid_count}")

# Check invalid dates (future or incorrect month/day)
total_dates = pd.to_datetime(data["release_date"], errors='coerce')
invalid_mask = (total_dates > pd.Timestamp.now()) | (total_dates.dt.month > 12) | (total_dates.dt.day > 31)
print(f"Invalid dates (future or wrong month/day): {invalid_mask.sum()}")
```

```python
# Numerical
rules = {
    'runtime': lambda x: (pd.to_numeric(x, errors='coerce') == 0),
    'budget': lambda x: pd.to_numeric(x, errors='coerce') == 0,
    'revenue': lambda x: pd.to_numeric(x, errors='coerce') == 0,
    'popularity': lambda x: pd.to_numeric(x, errors='coerce') == 0,
    'vote_average': lambda x: (pd.to_numeric(x, errors='coerce') == 0) | (pd.to_numeric(x, errors='coerce') > 10),
    'vote_count': lambda x: pd.to_numeric(x, errors='coerce') == 0,
    'id': lambda x: pd.to_numeric(x, errors='coerce') == 0,
}

# Create results DataFrame
results = []
for col, rule in rules.items():
    if col in data.columns:
        invalid_count = rule(data[col]).sum()
        total = data[col].notna().sum()
        results.append({
            'column': col,
            'invalid_count': invalid_count,
            'invalid_percent': round((invalid_count / total) * 100, 2)
        })

pd.DataFrame(results)
```
